```python
from IPython.core.magic import register_cell_magic
from IPython.display import HTML
import uuid

# only load the embed script once
_strudel_loader = """
<script>
if (!window.__strudel_embed_loaded__) {
    const s = document.createElement('script');
    s.src = "https://unpkg.com/@strudel/embed@latest";
    document.head.appendChild(s);
    window.__strudel_embed_loaded__ = true;
}
</script>
"""

@register_cell_magic
def strudel(line, cell):
    # wrap the pattern code inside an HTML comment
    code = "\n".join(cell.strip().splitlines())
    html_code = f"""
{_strudel_loader}
<style>
strudel-repl {{
    display: block;
    width: 100%;
}}
</style>
<div style="border:1px solid #ccc; border-radius:8px; padding:8px; width:100%;">
<strudel-repl>
<!--
{code}
-->
</strudel-repl>
</div>
"""
    return HTML(html_code)
```
